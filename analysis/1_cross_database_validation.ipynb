```python
# ============================================================
# Cross-database validation
# ============================================================
#
# This script compares mean ratings from the present study
# with established German normative databases:
#
#   1. Köper and Im Walde (2016)
#   2. Kanske and Kotz (2010)
#
# Pearson correlations are calculated for:
#   - concreteness
#   - arousal
#   - valence
#
```

```python
# ============================================================
# 1. Imports
# ============================================================

from pathlib import Path

import numpy as np
import pandas as pd
from scipy import stats


# ============================================================
# 2. Load datasets 
# ============================================================

study_df= pd.read_csv('../materials/MultiNorms_GER_summary_level_dataset.csv',index_col=0)

koeper_2016_df = pd.read_csv('PATH/TO/DATASET', sep="\t", index_col='Word')

kanske_2010_df= pd.read_csv('PATH/TO/DATASET', index_col='word')


# ============================================================
# 3. Check overlapping words
# ============================================================

koeper_overlap_words = (
    study_df.index
    .intersection(koeper_2016_df.index)
)

kanske_overlap_words = (
    study_df.index
    .intersection(kanske_2010_df.index)
)


print(
    f"Köper and Im Walde (2016): "
    f"{len(koeper_overlap_words)} overlapping nouns"
)

print(
    f"Kanske and Kotz (2010): "
    f"{len(kanske_overlap_words)} overlapping nouns"
)
```

```python
# ============================================================
# 4. Calculate correlations: Köper and Im Walde (2016)
# ============================================================

# 4.1 Filter 
koeper_2016_df= koeper_2016_df[koeper_2016_df.index.isin(study_df.index)]
study_df_koeper = study_df[study_df.index.isin(koeper_2016_df.index)]


# 4.2 Sort 
koeper_2016_df=koeper_2016_df.sort_index(ascending=True)
study_df_koeper=study_df_koeper.sort_index(ascending=True)


# 4.3 Check  
print(len(koeper_2016_df))
print(len(study_df_koeper))

if koeper_2016_df.index.equals(study_df_koeper.index):
    print("Indexes are the same.\n")
else:
    print("Indexes are different.\n")


# 4.4 Compute correlation 

res_corr = stats.pearsonr(koeper_2016_df["AbstConc"], study_df_koeper["concreteness_mean"])
ci_corr = res_corr.confidence_interval(confidence_level=0.95)

print(
    f"Concreteness: r = {res_corr.statistic:.2f}, "
    f"p = {res_corr.pvalue:.3g}, "
    f"95% CI [{ci_corr.low:.2f}, {ci_corr.high:.2f}]\n"
)

res_arousal = stats.pearsonr(koeper_2016_df["Arou"], study_df_koeper["arousal_mean"])
ci_arou = res_arousal.confidence_interval(confidence_level=0.95)

print(
    f"Arousal: r = {res_arousal.statistic:.2f}, "
    f"p = {res_arousal.pvalue:.3g}, "
    f"95% CI [{ci_arou.low:.2f}, {ci_arou.high:.2f}]\n"
)

res_valence = stats.pearsonr(koeper_2016_df["Val"], study_df_koeper["valence_mean"])
ci_val = res_valence.confidence_interval(confidence_level=0.95)

print(
    f"Valence: r = {res_valence.statistic:.2f}, "
    f"p = {res_valence.pvalue:.3g}, "
    f"95% CI [{ci_val.low:.2f}, {ci_val.high:.2f}]\n"
)
```

```python
# ============================================================
# 5. Calculate correlations: Kanske and Kotz (2010)
# ============================================================

# 5.1 Filter 
kanske_2010_df= kanske_2010_df[kanske_2010_df.index.isin(study_df.index)]
study_df_kanske= study_df[study_df.index.isin(kanske_2010_df.index)]


# 5.2 Sort
kanske_2010_df=kanske_2010_df.sort_index(ascending=True)
study_df_kanske=study_df_kanske.sort_index(ascending=True)


# 5.3 Check  
print(len(kanske_2010_df))
print(len(study_df_kanske))


#5.4 Exclude words used for cross-validation quality check 

words_qc= ['Absatz', 'Anzahl', 'Bildung', 'Dank', 'Ehe', 'Einfahrt', 'Folter', 'Holz',
 'Kippe', 'Kraft', 'Leid', 'Leiter', 'Moment', 'Name', 'Ordner', 'Panik', 'Papier',
 'Schiff', 'Stahl', 'Steuer', 'Streifen', 'Treue', 'Tuch', 'Vorteil', 'Wahrheit']

# Remove QC words from both datasets
kanske_2010_df = kanske_2010_df.drop(index=words_qc, errors="ignore")
study_df_kanske = study_df_kanske.drop(index=words_qc, errors="ignore")

# check length after exclusion  
print(len(kanske_2010_df))
print(len(study_df_kanske))

# check if databases have same index 
if kanske_2010_df.index.equals(study_df_kanske.index):
    print("Indexes are the same.\n")
else:
    print("Indexes are different.\n")


# 5.5 Compute correlation 

res_corr = stats.pearsonr(
    kanske_2010_df["concreteness_mean"],
    study_df_kanske["concreteness_mean"]
)
ci_corr = res_corr.confidence_interval(confidence_level=0.95)

print(
    f"Concreteness: r = {res_corr.statistic:.2f}, "
    f"p = {res_corr.pvalue:.3g}, "
    f"95% CI [{ci_corr.low:.2f}, {ci_corr.high:.2f}]\n"
)

res_arousal = stats.pearsonr(
    kanske_2010_df["arousal_mean"],
    study_df_kanske["arousal_mean"]
)
ci_arou = res_arousal.confidence_interval(confidence_level=0.95)

print(
    f"Arousal: r = {res_arousal.statistic:.2f}, "
    f"p = {res_arousal.pvalue:.3g}, "
    f"95% CI [{ci_arou.low:.2f}, {ci_arou.high:.2f}]\n"
)

res_valence = stats.pearsonr(
    kanske_2010_df["valence_mean"],
    study_df_kanske["valence_mean"]
)
ci_val = res_valence.confidence_interval(confidence_level=0.95)

print(
    f"Valence: r = {res_valence.statistic:.2f}, "
    f"p = {res_valence.pvalue:.3g}, "
    f"95% CI [{ci_val.low:.2f}, {ci_val.high:.2f}]\n"
)
```
