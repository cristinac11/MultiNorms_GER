```python
# Hierarchical cluster analysis

# This notebook performs a hierarchical cluster analysis of the three-factor scores obtained from the exploratory factor analysis. 
# Euclidean distances and Ward’s linkage method are used to identify groups of words with similar factor profiles. 
# A dendrogram is used to visualise the hierarchical structure, followed by an agglomerative clustering solution with four clusters. 
# The resulting cluster assignments are added to the reduced factor-score dataset.
```

```python
# ============================================================
# 1. Imports
# ============================================================

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering, KMeans
from sklearn.metrics import cohen_kappa_score
from sklearn.decomposition import PCA
from scipy.cluster.hierarchy import linkage, dendrogram
import matplotlib.pyplot as plt

# ============================================================
# 2. Load dataset
# ============================================================

df = pd.read_csv('MultiNorms_GER_reduced_df_efa_3factors',  index_col=0)

# ============================================================
# 3. Dendogram
# ============================================================

linkage_matrix = linkage(df, method='ward', metric='euclidean')

# Plot the dendrogram
plt.figure(figsize=(12, 6))
dendrogram(linkage_matrix, truncate_mode='lastp', leaf_rotation=90., leaf_font_size=8., p=20, show_contracted=True)
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Data Points')
plt.ylabel('Distance')
plt.show()
```

```python
# ============================================================
# 4. Perform hierarchical clustering with 4 clusters
# ============================================================

final_model= AgglomerativeClustering(n_clusters=4, linkage='ward')
labels = final_model.fit_predict(df)

# ============================================================
# 5. Save dataset with cluster labels 
# ============================================================

df['cluster']=labels
df.to_csv("MultiNorms_GER_reduced_df_efa_3factors_clusters.csv", encoding='utf-8-sig')
```
